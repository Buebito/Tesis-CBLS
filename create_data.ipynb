# Crea el csv

```python
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
eth1 = pd.read_csv('data/raw/Binance_ETHUSDT_2021_minute.csv', parse_dates=['date'])
eth2 = pd.read_csv('data/raw/Binance_ETHUSDT_2022_minute.csv', parse_dates=['date'])
eth3 = pd.read_csv('data/raw/Binance_ETHUSDT_2023_minute.csv', parse_dates=['date'])
eth4 = pd.read_csv('data/raw/Binance_ETHUSDT_2024_minute.csv', parse_dates=['date'])
eth5 = pd.read_csv('data/raw/Binance_ETHUSDT_2025_minute.csv', parse_dates=['date'])

btc1 = pd.read_csv('data/raw/Binance_BTCUSDT_2021_minute.csv', parse_dates=['date'])
btc2 = pd.read_csv('data/raw/Binance_BTCUSDT_2022_minute.csv', parse_dates=['date'])
btc3 = pd.read_csv('data/raw/Binance_BTCUSDT_2023_minute.csv', parse_dates=['date'])
btc4 = pd.read_csv('data/raw/Binance_BTCUSDT_2024_minute.csv', parse_dates=['date'])
btc5 = pd.read_csv('data/raw/Binance_BTCUSDT_2025_minute.csv', parse_dates=['date'])


eth = pd.concat([eth5, eth4, eth3, eth2, eth1], ignore_index=True)
btc = pd.concat([btc5, btc4, btc3, btc2, btc1], ignore_index=True)

eth = eth.rename(columns={'close': 'ETH', 'high': 'ETH_HIGH', 'low': 'ETH_LOW'})
btc = btc.rename(columns={'close': 'BTC', 'high': 'BTC_HIGH', 'low': 'BTC_LOW'})

btc = btc[['date', 'BTC','BTC_HIGH', 'BTC_LOW']]
eth = eth[['date', 'ETH','ETH_HIGH', 'ETH_LOW']]

btc['date'] = pd.to_datetime(btc['date'], format="%Y-%m-%d %H:%M:%S", errors="coerce")
eth['date'] = pd.to_datetime(eth['date'], format="%Y-%m-%d %H:%M:%S", errors="coerce")

btc = btc.dropna(subset=['date'])
eth = eth.dropna(subset=['date'])

btc = btc.sort_values(by='date').reset_index(drop=True)
eth = eth.sort_values(by='date').reset_index(drop=True)

merged = pd.merge(btc, eth, on='date', how='inner')
merged.to_csv('data/processed/historic_prices.csv', index=False)
```

# Creal el Log csv

```python
def generate_log_csv(src_csv: str | Path, dst_csv: str | Path = "prices_log.csv"):

    df = pd.read_csv(src_csv, parse_dates=["date"]).set_index("date").sort_index()
    price_cols = [c for c in df.columns if c.endswith(("BTC", "ETH", "_HIGH", "_LOW"))]
    df[price_cols] = np.log(df[price_cols].astype(float))
    df.to_csv(dst_csv, index=True)

generate_log_csv("data/processed/historic_prices.csv", "data/processed/historic_prices_log.csv")
```
